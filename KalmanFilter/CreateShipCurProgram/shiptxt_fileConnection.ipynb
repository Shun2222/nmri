```python
%matplotlib inline
import datetime
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import os.path as osp
import re
```

```python
def file_connection(filenames, save_filepath):
    combined_text = []
    count = 0
    for fn in filenames:
        try:
            with open(fn, 'r') as file:
                lines = file.readlines()
                combined_text.extend(lines)
                count += 1
        except FileNotFoundError:
            None
            #print(f"{fn} is not found.")
    print(f"{count}/{len(filenames)} files are existed")
    # 結合されたテキストを新しいファイルに書き込み
    with open(save_filepath, 'w') as combined_file:
        combined_file.writelines(combined_text)
    print(f"Saved connected file in {save_filepath}")
```

```python
#filename = "./data/2015-1230-31/中春/中春丸FileOut20160919"
#file_names = [f"{filename}-{i:04}.txt" for i in range(0, 2360, 10)]


path = "E:\shunsukeE\data\shiplog"
files = os.listdir(path)
dirs = [f for f in files if os.path.isdir(os.path.join(path, f))]

patterns = []
for month in [i for i in range(1, 13)]:
    for day in [i for i in range(1, 32)]:
        if (month == 2 and day > 28): continue
        if (month < 8 and month % 2 == 0 and day == 31): continue
        if (month >= 8 and month % 2 != 0 and day == 31): continue
        patterns.append('(\w+){month}{day}-')

for d in dirs:
    path2 = osp.join(path, d)
    files = os.listdir(path2)
    filenames = [f for f in files if os.path.isfile(osp.join(path2, f))]
    for pattern in patterns:
        target = []
        for f in filenames:
            if re.match(pattern, f):
                f_path = osp.join(path, d, f)
                target.append(f_path)
            else:
                continue
        if target:
            file_connection(target, target[0][:-9]+'.txt')
        else:
            print(f'target is none in {d}')
```

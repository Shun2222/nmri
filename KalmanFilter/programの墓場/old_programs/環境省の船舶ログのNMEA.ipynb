```python
import csv

# AIS2 
path_ais = [ './data/test/SUNNYMARSFileOut20160101-0000.txt'] # 0-14時

datas = []
for i, p in enumerate(path_ais):
    i = i+4
    with open(p, "r", encoding="cp932") as f:
        reader = csv.reader(f, delimiter=",", doublequote=True, lineterminator="\n", quotechar='"', skipinitialspace=False)
        for row in reader:
            datas.append(row)
ais = pd.DataFrame(datas)
#column_names = ["JST", "Institution", "DataType"]
#ais.columns = column_names
ais.head()
```
